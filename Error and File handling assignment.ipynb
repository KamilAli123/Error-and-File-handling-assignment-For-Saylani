### Q4 
Ek file `numbers.txt` banayein aur us mein 1 se 5 tak numbers line-by-line write karein.

```python
with open("numbers.txt", "a") as file:
    for a in range(1,6):
        file.write(str(f"{a}\n"))
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
with open("numbers.txt", "r") as file:
    data = file.readlines()
    total = 0
    for a in data:
        total += int(a)
    print(total)
```

### Q6

```python
with open("numbers.txt", "r") as file:
    data = file.readlines()
    total_lines = len(data)
    print(total_lines)
```

### Q7 (Medium)
`with open()` statement ka istemal karte hue ek file `data.txt` read karein aur explain karein ki `with` ka kya faida hai.

```python
with open("data.txt", "r") as file:
    print(file.read())

#Advantages Of Using "With" Statement Are:
#You don't need to use .close() method to close the file because "With" Statement automatically closes the file after finishing
#"With" Statement reduces the size of the code because you no longer need to use .close() to close files anymore while using "With" Statement
#with using "With" Statement the file with automatically close in case of crash or error which would not happen if "With" Statement was not being used
```

### Q8 (Medium)
Ek file `sample.txt` se read karke uski saari lines ko ek Python list mein convert karein (`readlines()` ka use karke).

```python
with open("sample.txt", "r+") as file:
    data = file.readlines()
    list1 = []
    for a in data:
        list1.append(str(a))
    print(list1)
```

### Q9 
Ek file `source.txt` ka data read karke `destination.txt` file mein copy karein.

```python
with open("source.txt", "r") as source, open("destination.txt", "w+") as dest:
    data = source.read()
    dest.write(data)
    dest.seek(0)
    print(dest.read())
```

### Q10 
`os` module ka use karke check karein ki kya `test.txt` file exist karti hai ya nahi. Agar karti hai toh usay delete kar dein.

```python
import os

if os.path.exists("test.txt"):
    print("File test.txt Exists")
    os.remove("test.txt")
    print("File test.txt Removed")
```

### Q11 
Aisa program likhein jo 10 ko 0 se divide kare aur `try-except` block ka use karke `ZeroDivisionError` ko handle kare.

```python
try:
    calc = 10 / 0
    print(calc)
except ZeroDivisionError:
    print("A Number Cannot Be Divided By Zero")
```

### Q13 (Easy)

### Q14

### Q15 
Ek list `my_list = [10, 20, 30]` banayein. Invalid index (e.g., index 5) access karein aur `IndexError` catch karein.

```python
my_list = [10, 20, 30]
try:
    my_list[5]
except IndexError:
    print("The Given Index Does Not Exist")
```

### Q16 
Ek dictionary `student = {"name": "Ali"}` banayein. Non-existent key (`"age"`) access karein aur `KeyError` handle karein.

```python
student = {"name" : "Ali"}

try:
    student["age"]
except KeyError:
    print("The Given Key Does Not Exist In The Dictionary")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
age = int(input("Enter Age:"))

if age < 18:
    raise ValueError("Age Cannot Be Less Then 18 To Enter")
else:
    print("Your Welcome In")
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
try:
    file = open("Script.txt", "r")
except FileNotFoundError:
    print("File Script.txt Does Not Exist")
finally:
    file.close()
```
